```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from finrl.meta.preprocessor.yahoodownloader import YahooDownloader
from finrl.meta.preprocessor.preprocessors import FeatureEngineer, data_split
from finrl import config_tickers
from finrl.config import INDICATORS
```

```python
from finrl import config
from finrl import config_tickers
```

```python
import os
if not os.path.exists("./" + config.DATA_SAVE_DIR):
    os.makedirs("./" + config.DATA_SAVE_DIR)
```

```python
print(config_tickers.DOW_30_TICKER)
```

```python
# Tải dữ liệu từ 2009 đến cuối năm 2025
df = YahooDownloader(start_date = '2009-01-01',
                     end_date = '2025-12-20',
                     ticker_list = config_tickers.DOW_30_TICKER).fetch_data()

print(f"Dữ liệu mới có {df.shape[0]} dòng.")
```

```python
# 1. Import các thư viện cần thiết cho bước này (nếu chưa import)
from finrl.meta.preprocessor.preprocessors import FeatureEngineer, data_split
from finrl.config import INDICATORS

# 2. Khởi tạo Kỹ sư xử lý dữ liệu
print("Đang khởi tạo FeatureEngineer...")
fe = FeatureEngineer(
                    use_technical_indicator=True,
                    tech_indicator_list = INDICATORS, # Gồm: macd, boll_ub, boll_lb, rsi_30, cci_30, dx_30, close_30_sma, close_60_sma
                    use_vix=True,        # Thêm chỉ số biến động VIX
                    use_turbulence=True, # Thêm chỉ số hỗn loạn thị trường
                    user_defined_feature = False)

# 3. Thực hiện tính toán (Bước này mất khoảng 1-2 phút)
processed = fe.preprocess_data(df)

print("Đã thêm các chỉ báo thành công!")
print(processed.head())
```

```python
# Lấy danh sách toàn bộ mã cổ phiếu và toàn bộ ngày
list_ticker = processed["tic"].unique().tolist()
list_date = list(pd.date_range(processed['date'].min(),processed['date'].max()).astype(str))
combination = pd.DataFrame({'date':list_date})

# Tạo một bảng khung (skeleton) chứa đầy đủ ngày cho từng mã
processed_full = pd.DataFrame()
for ticker in list_ticker:
    temp = combination.copy()
    temp['tic'] = ticker
    processed_full = pd.concat([processed_full, temp], axis=0)

# Ghép dữ liệu đã tính toán vào bảng khung
processed = processed_full.merge(processed, on=['date','tic'], how='left')

# Sắp xếp lại
processed = processed.sort_values(['date','tic'])

# Điền dữ liệu thiếu (fillna)
# Logic: Dùng dữ liệu ngày hôm trước điền cho ngày hôm sau (ffill)
processed = processed.fillna(0) # Bước phòng ngừa cuối cùng

print("Dữ liệu đã được làm sạch và lấp đầy.")
print(f"Tổng số dòng sau khi xử lý: {processed.shape[0]}")
```

```python
# Chia tập Train: Từ 2009-01-01 đến 2024-01-01
train = data_split(processed, '2009-01-01','2024-01-01')

# Chia tập Trade: Từ 2024-01-01 đến 2025-12-20
trade = data_split(processed, '2024-01-01','2025-12-20')

print(f"Độ dài tập Train: {len(train)}")
print(f"Độ dài tập Trade: {len(trade)}")
```

```python
train.to_csv('train_data.csv')
trade.to_csv('trade_data.csv')

print("Đã lưu xong! Hãy tải 2 file train_data.csv và trade_data.csv về máy.")
```
